Q4. Track passengers on a bus:
passengers = ["Ali", "Zara", "Ahmed"]
Insert a new passenger "Sara" at position 2
Remove passesnger who got off "Zara"
Reverse the order to show who entered first

```python
passengers = ["Ali", "Zara", "Ahmed"]

# Insert Sara at position 2 (index 1)
passengers.insert(1, "Sara")

# Remove Zara
passengers.remove("Zara")

# Reverse list
passengers.reverse()

print(passengers)
```

Q5. text = 'hello python world'
Replace spaces with underscores in the string (e.g., "hello python world" → "hello_python_world").
Split the modified string (e.g., "hello_python_world" → ['hello', 'python', 'world']).
Join the list elements with a hyphen (-) between them (e.g., ['hello', 'python', 'world'] → "hello-python-world").
Print all three results
Hint: Use string methods

```python
text = "hello python world"

# Replace spaces with underscores
modified = text.replace(" ", "_")

# Split using underscore
words = modified.split("_")

# Join with hyphen
joined = "-".join(words)

print("Modified:", modified)
print("Split:", words)
print("Joined:", joined)
```

Q7. You have a product code that follows the format "SKU-12345-XYZ".
Find the first position of the dash (-).
Count the number of dashes (-) in the product code.
Remove the "SKU-" prefix and "-XYZ" suffix from the product code, Extract the numeric part 12345
Hint: Use String Methods


```python
code = "SKU-12345-XYZ"

# Find first dash
print("First dash position:", code.find("-"))

# Count dashes
print("Number of dashes:", code.count("-"))

# Extract numeric part
number = code.replace("SKU-", "").replace("-XYZ", "")

print("Numeric Part:", number)
```

Q9. text = " Data Analysis with Python is very powerful "
Perform the following tasks:
Remove extra spaces using strip().
Convert the string into title case using title().
Replace "Powerful" with "Important".
Find the position of the word "Python".

```python
text = " Data Analysis with Python is very powerful "

# Remove spaces
text = text.strip()

# Title case
title_text = text.title()

# Replace Powerful with Important
new_text = title_text.replace("Powerful", "Important")

# Find Python
position = new_text.find("Python")

print("After strip:", text)
print("Title Case:", title_text)
print("After Replace:", new_text)
print("Position of Python:", position)
```

Q10. Given the list:
data = [12, 45, 2, 67, 34, 89, 23, 90]
Perform the following tasks:
Sort the list in ascending order.
Remove the highest value from the list.
Insert the value 50 at index 3.".
Finally, reverse the updated list and display the result.

```python
data = [12, 45, 2, 67, 34, 89, 23, 90]

# Sort
data.sort()

# Remove highest value
data.remove(max(data))

# Insert 50 at index 3
data.insert(3, 50)

# Reverse
data.reverse()

print(data)
```
