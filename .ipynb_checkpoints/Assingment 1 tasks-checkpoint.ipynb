Q4.

Check the type and convert: (num1 = 10, num2 = 5.6, num3 = 4.43, num4='23')

Integer to float

Float to integer

Float to String

String to Integer

```python
num1 = 10
num2 = 5.6
num3 = 4.43
num4 = '23'

# Check data types
print("Type of num1:", type(num1))
print("Type of num2:", type(num2))
print("Type of num3:", type(num3))
print("Type of num4:", type(num4))

# Type conversions

# Integer to Float
int_to_float = float(num1)
print("Integer to Float:", int_to_float)

# Float to Integer
float_to_int = int(num2)
print("Float to Integer:", float_to_int)

# Float to String
float_to_string = str(num3)
print("Float to String:", float_to_string)

# String to Integer
string_to_int = int(num4)
print("String to Integer:", string_to_int)
```

Add a and b

Find modulus of c divided by d

```python
# Given values
a = 2
b = 1
c = 6
d = 3

# Operations
print("Add a and b =", a + b)
print("Subtract c and a =", c - a)
print("Multiply d and b =", d * b)
print("Divide c and d =", c / d)
print("Modulus of c divided by d =", c % d)
print("Square of (d + a) =", (d + a) ** 2)
```

Q8.

a = '2' b = '11' c = 6.3 Convert a to integer then add to c Convert b to float then multiply by c Convert c to integer then add to c

```python
# Given values
a = '2'
b = '11'
c = 6.3

# Convert a to integer then add to c
print("a + c =", int(a) + c)

# Convert b to float then multiply by c
print("b * c =", float(b) * c)

# Convert c to integer then add to c
print("int(c) + c =", int(c) + c)
```

Q9. 

You had 12 books, and you gave 4 to your friend. How many books do you have now? Store books in a variable called books. Print final output in same variable

```python
# Store books
books = 12

# Give away 4 books
books = books - 4

# Print remaining books
print("Books left:", books)
```
