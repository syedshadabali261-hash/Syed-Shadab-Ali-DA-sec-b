```python
sing python csv module add student grades column according to percentages, write data in new csv file including grades column:


Grade
A > 80%–100%

B > 70%–79%

C > 60%–69%

D > 50%–59%

F > below 50%
```

```python
import csv

input_file = "student data - student data.csv"
output_file = "student_data_with_grades.csv"

def calculate_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

with open(input_file, "r", newline="") as infile:
    reader = csv.DictReader(infile)

    fieldnames = reader.fieldnames + ["Grade"]

    with open(output_file, "w", newline="") as outfile:
        writer = csv.DictWriter(outfile, fieldnames=fieldnames)
        writer.writeheader()

        for row in reader:
            row["Grade"] = calculate_grade(float(row["Percentage"]))
            writer.writerow(row)

print("Done! Grade column added.")
```
