Question 1
Employee Data Bank.
Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri- butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each instance. Then do some updations with it's methods.


```python
class Employee:

    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("\n--- Employee Information ---")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    def update_year_of_experience(self, years):
        self.year_of_experience += years
        print("Updated Experience:", self.year_of_experience)

    def increase_salary(self, amount):
        self.salary += amount
        print("Updated Salary:", self.salary)


# Employee 101
employee101 = Employee(
    "Ali Khan",
    "03001234567",
    "Karachi",
    50000,
    "Software Developer",
    2022,
    "BS Computer Science",
    2
)

# Print four attributes individually
print("Name:", employee101.name)
print("Salary:", employee101.salary)
print("Post:", employee101.post)
print("Education:", employee101.education)

# Call all methods
employee101.describe_employee()

employee101.increase_salary(10000)

employee101.update_year_of_experience(1)


# Creating 10 different employees
employee1 = Employee("Ahmed", "03001111111", "Karachi", 45000, "Developer", 2023, "BSCS", 1)
employee2 = Employee("Sara", "03002222222", "Lahore", 55000, "Designer", 2021, "BSIT", 3)
employee3 = Employee("John", "03003333333", "Islamabad", 60000, "Manager", 2020, "MBA", 5)
employee4 = Employee("Ayesha", "03004444444", "Karachi", 50000, "HR Officer", 2022, "BBA", 2)
employee5 = Employee("Bilal", "03005555555", "Hyderabad", 48000, "Programmer", 2023, "BSCS", 1)
employee6 = Employee("Hina", "03006666666", "Multan", 52000, "Accountant", 2022, "B.Com", 2)
employee7 = Employee("Usman", "03007777777", "Karachi", 70000, "Team Lead", 2019, "BSCS", 6)
employee8 = Employee("Zara", "03008888888", "Lahore", 58000, "Web Developer", 2021, "BSIT", 4)
employee9 = Employee("Hamza", "03009999999", "Quetta", 47000, "Tester", 2023, "BSCS", 1)
employee10 = Employee("Maham", "03001010101", "Islamabad", 65000, "Project Manager", 2020, "MBA", 5)



employees = [
    employee1,
    employee2,
    employee3,
    employee4,
    employee5,
    employee6,
    employee7,
    employee8,
    employee9,
    employee10
]


print("\n\n===== ALL EMPLOYEES =====")

for employee in employees:
    employee.describe_employee()


print("\n===== UPDATES =====")

employee1.increase_salary(5000)
employee2.update_year_of_experience(1)
employee3.increase_salary(10000)
employee4.update_year_of_experience(2)
employee5.increase_salary(7000)
```
