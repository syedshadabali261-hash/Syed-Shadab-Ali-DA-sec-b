```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('factbook.db')
print('Database connected successfully.')
```

## Q1. List all countries where the area is greater than 5000, along with their population.


```python
query = '''
SELECT name AS Country, area AS Area, population AS Population
       FROM facts
       WHERE area > 5000
       ORDER BY area DESC;
'''
result = pd.read_sql_query(query, conn)
result
```

## Q2. List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%.


```python
query = '''
SELECT c.name AS Capital_City, f.name AS Country,
              f.death_rate AS Death_Rate, f.birth_rate AS Birth_Rate
       FROM cities c
       JOIN facts f ON c.facts_id = f.id
       WHERE c.capital = 1
         AND f.death_rate > 2
         AND f.birth_rate < 20;
'''
result = pd.read_sql_query(query, conn)
result
```

## Q3. List the cities with a population growth rate between 1% and 5%, sorted in descending order by population growth.


```python
query = '''
SELECT c.name AS City, f.name AS Country,
              f.population_growth AS Population_Growth
       FROM cities c
       JOIN facts f ON c.facts_id = f.id
       WHERE f.population_growth BETWEEN 1 AND 5
       ORDER BY f.population_growth DESC;
'''
result = pd.read_sql_query(query, conn)
result
```

## Q4. Find countries that start with "A" and do not have any cities listed in the cities table. Use LEFT JOIN.


```python
query = '''
SELECT f.name AS Country, c.name AS City
       FROM facts f
       LEFT JOIN cities c ON f.id = c.facts_id
       WHERE f.name LIKE 'A%'
         AND c.id IS NULL;
'''
result = pd.read_sql_query(query, conn)
result
```

## Q5. List all capital cities and their corresponding death rate using a LEFT JOIN.


```python
query = '''
SELECT c.name AS Capital_City, f.death_rate AS Death_Rate
       FROM facts f
       LEFT JOIN cities c
         ON f.id = c.facts_id AND c.capital = 1
       WHERE c.id IS NOT NULL
       ORDER BY c.name;
'''
result = pd.read_sql_query(query, conn)
result
```
