## Question 1 — Check Duplicate Records

```python
import pandas as pd
df = pd.read_csv("cars - cars.csv")
df.duplicated().sum()
```

## Question 2 — Drop Duplicate Records

```python
df_no_duplicates = df.drop_duplicates()
print("Original records:", len(df))
print("Records after dropping duplicates:", len(df_no_duplicates))
df_no_duplicates.head()
```

## Question 3 — Check Missing or Null Values

```python
df.isnull().sum()
```

## Question 4 — Drop Missing or Null Values

```python
df_no_missing = df.dropna()
print("Records before dropping missing values:", len(df))
print("Records after dropping missing values:", len(df_no_missing))
df_no_missing.head()
```

## Question 5 — Top 10 Average MSRP per Car Brand

```python
top_10_brands = (
    df.groupby("Make")["MSRP"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
      .round(0)
)

top_10_brands
```

## Question 6 — Average MSRP per Year

```python
average_msrp_year = (
    df.groupby("Year")["MSRP"]
      .mean()
      .round(0)
)

average_msrp_year
```

## Question 7 — Average Engine HP and MPG per Year

Average MPG is calculated as:

**(City MPG + Highway MPG) / 2**


```python
df["Average MPG"] = (df["city mpg"] + df["highway MPG"]) / 2

year_summary = (
    df.groupby("Year")
      .agg({
          "Engine HP": "mean",
          "Average MPG": "mean"
      })
      .round(0)
)

year_summary
```

## Question 8 — Percentage of Compact, Midsize, or Large Cars

```python
size_percentage = (
    df["Vehicle Size"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

size_percentage
```

## Question 9 — Most Common Vehicle Styles

```python
vehicle_styles = df["Vehicle Style"].value_counts()
vehicle_styles
```

## Question 10 — Brands with Most Vehicles Priced at $40,000+ and Popularity at Least 1000

```python
filtered_cars = df[
    (df["MSRP"] >= 40000) &
    (df["Popularity"] >= 1000)
]
brand_counts = filtered_cars["Make"].value_counts()
brand_counts
```
