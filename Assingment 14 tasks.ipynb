# Assignment 14 – Employee Salary Analysis

This notebook uses the provided `database.sqlite` database and the `Salaries` table.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("database.sqlite")

df = pd.read_sql("SELECT * FROM Salaries", conn)

df.head()
```

## Q1. Average TotalPayBenefits for Each JobTitle

Exclude `Not Provided` and `Not provided`.

```python
q1 = (
    df[~df["JobTitle"].isin(["Not Provided", "Not provided"])]
    .groupby("JobTitle", as_index=False)["TotalPayBenefits"]
    .mean()
    .rename(columns={"TotalPayBenefits": "AverageTotalPayBenefits"})
    .sort_values("AverageTotalPayBenefits", ascending=False)
)

q1
```

## Q2. JobTitle and OvertimePay for Employees with OvertimePay > 50000

```python
df["OvertimePay"] = pd.to_numeric(df["OvertimePay"], errors="coerce")

q2 = df.loc[
    df["OvertimePay"] > 50000,
    ["JobTitle", "OvertimePay"]
].sort_values("OvertimePay", ascending=False)

q2
```

## Q3. Total Number of Unique Employees

Exclude employee names `Not Provided` and `Not provided`.

```python
q3 = df[
    ~df["EmployeeName"].isin(["Not Provided", "Not provided"])
]["EmployeeName"].nunique()

print("Total number of unique employees:", q3)
```

## Q4. Top 10 Highest Paid Employees

```python
q4 = (
    df.sort_values("TotalPayBenefits", ascending=False)
      [["EmployeeName", "JobTitle", "TotalPayBenefits"]]
      .head(10)
)

q4
```

## Q5. All Employees Ordered by TotalPayBenefits in Descending Order

```python
q5 = df.sort_values(
    "TotalPayBenefits",
    ascending=False
)[["EmployeeName", "JobTitle", "TotalPayBenefits"]]

q5
```
